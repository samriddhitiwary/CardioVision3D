# CardioTwin Final Evaluation and Model Freeze

This notebook freezes the ML system. Modeling decisions from Notebook 04 are locked. Notebook 05 performs locked-configuration repeated stratified cross-validation, evaluates frozen thresholds and sigmoid calibration, trains deployment pipelines on all 303 rows, saves metadata/schema/model card, and verifies reload inference.

This is not an external test set and not an independent clinical cohort.

```python
from __future__ import annotations

import json
import sys
from pathlib import Path

import pandas as pd

cwd = Path.cwd().resolve()
for candidate in [cwd, *cwd.parents]:
    for maybe_ml in [candidate, candidate / "ml"]:
        if (maybe_ml / "src" / "models" / "final_evaluation.py").exists():
            if str(maybe_ml) not in sys.path:
                sys.path.insert(0, str(maybe_ml))
            break
    else:
        continue
    break

from src.models.final_evaluation import FORBIDDEN_COLUMNS, TARGETS, find_project_root, load_inputs, run_final_evaluation

PROJECT_ROOT = find_project_root()
X, y, feature_metadata, numeric_features, categorical_features, baseline_summary, baseline_oof, baseline_overfit = load_inputs(PROJECT_ROOT)

assert len(X) == len(y) == 303
assert X.index.equals(y.index)
assert list(y.columns) == TARGETS
assert FORBIDDEN_COLUMNS.isdisjoint(X.columns), "Leakage columns found in X"
assert X.columns.is_unique
assert not any(str(c).lower().startswith("unnamed") for c in X.columns)
for target in TARGETS:
    assert set(y[target].unique()).issubset({0, 1})

print(f"Project root: {PROJECT_ROOT}")
print(f"X shape: {X.shape}; y shape: {y.shape}")
print(f"Numeric predictors: {len(numeric_features)}")
print(f"Categorical/binary predictors: {len(categorical_features)}")
print("Strict safety checks passed.")
```

## Run Locked Final Evaluation and Freeze

All preprocessing, feature selection, calibration, and fitting occur inside training folds during locked repeated CV. Deployment models are trained on all rows only after validation is complete.

```python
results = run_final_evaluation(PROJECT_ROOT)
print("Execution seconds:", results["execution_seconds"])
print("Saved models:")
print(json.dumps(results["saved_model_paths"], indent=2))
print("Warnings:", results["warnings"])
```

## Final Summary Tables

```python
final_summary = results["final_summary"]
threshold_comparison = results["threshold_comparison"]
calibration_metrics = results["calibration_metrics"]
stability_summary = results["stability_summary"]
notebook04_vs_final = results["notebook04_vs_final"]
cad_ensemble = results["cad_ensemble"]

reporting = final_summary[(final_summary["probability_type"] == "sigmoid") & (final_summary["threshold_name"] == "frozen")].copy()
cols = [
    "target", "accuracy_mean", "precision_mean", "recall_mean", "specificity_mean", "f1_mean",
    "balanced_accuracy_mean", "mcc_mean", "roc_auc_mean", "pr_auc_mean", "brier_mean",
    "roc_auc_std", "f1_std",
]
display(reporting[cols])

threshold_wide = threshold_comparison.pivot(index="target", columns="threshold_name", values=["f1", "recall", "specificity"])
display(threshold_wide)

display(calibration_metrics)
display(stability_summary)
display(notebook04_vs_final)
display(cad_ensemble)
print("All reported final validation predictions were generated out-of-fold under locked-configuration repeated stratified cross-validation.")
```

```python
expected_final = [
    "final_cv_metrics.csv",
    "final_summary.csv",
    "final_oof_predictions.csv",
    "threshold_comparison.csv",
    "calibration_metrics.csv",
    "stability_summary.csv",
    "notebook04_vs_final.csv",
    "cad_ensemble_comparison.csv",
]
final_dir = results["final_dir"]
for name in expected_final:
    path = final_dir / name
    assert path.exists(), f"Missing final artifact: {path}"
    print(name, pd.read_csv(path).shape)
for target in TARGETS:
    path = results["models_dir"] / f"{target}_pipeline.joblib"
    assert path.exists(), f"Missing model: {path}"
    print(path)
for path in [results["metadata_path"], results["input_schema_path"], results["model_card_path"]]:
    assert Path(path).exists(), f"Missing artifact: {path}"
    print(path)
print("Reload smoke test:")
display(results["reload_smoke_test"])
```
